# Pandas series


A series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

Series are kept vertically.

### Creating series

There are multiple ways to create a pandas series.

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c'] # List
my_list = [10, 20, 30] # List 
arr = np.array([10, 20, 30]) # Array
d = { 1:10, 2:20, 3:30 } # Dictionary
```

```python
pd.Series(my_list)
```

We can also set custom labels. 

```python
pd.Series(my_list, index = labels)
```

```python
pd.Series(arr) # We can't use 2D arrays because the Series in 1D itself.
```

```python
pd.Series(d) # The keys of dictionary are assigned automatically as index for the Series
```

```python
dd = { 'a':10, 'b':20, 'c':30 } # Dictionary
pd.Series(dd)
```

If we combined multiple series then it will become **dataframes**.
