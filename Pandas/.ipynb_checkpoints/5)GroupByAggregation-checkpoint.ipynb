```python
import numpy as np 
import pandas as pd 
```

## Group By

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
df
```

```python
# Group by Category and calculate the sum of Sales
df.groupby('Category')
```

Group by gives an object and we can't see it directly

```python
cat = df.groupby('Category')
for i, v in cat:
    print(i)
    print(v)
```

```python
cat = df.groupby('Category')['Sales'].sum()
cat
```

```python
# Group by store and calculate the sum of sales
df.groupby('Store')['Sales'].sum()
```

When we group-by, we do it in categorical data.

Therea are two types of data:
- qualitative data [categorial data like male,female, stores, etc] -> used for group-by
- quantitative data [quantity, sales, etc which have numbers]

```python
# Group by multiple columns 
# Group by category and store
df.groupby(['Category' ,'Store'])['Sales'].sum()
```

## Aggregation 

```python
df['Sales'].mean() # similary there are median, minimum, maximum, standard deviation, count, etc.
```

```python
# We can use agg() method to pass multiple aggregate functions and have multiple results at once
df['Sales'].agg(['mean', 'min', 'max']) 
```
