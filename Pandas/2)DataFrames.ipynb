```python
import numpy as np
import pandas as pd 
```

### Creating a DataFrame

We can create DataFrame using arrays, dictionaries, and list.

```python
data = {
    'Name' : ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New york', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
} # This is a dictionary

df = pd.DataFrame(data)
```

```python
df
```

```python
# Dataframe using List
data_list = [
    ['John', 28, 'New york', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
```

```python
df2
```

In dictionary we can get the name of columns directly, but when using lists, we will get default 0,1,2,... in both rows and columns.

We can assign columns in the DataFrame using another list.

```python
columns = ['Name', 'Age', 'City', 'Salary']
```

```python
df2 = pd.DataFrame(data_list, columns = columns)
df2
```

```python
df2['Name'] # Selecting a single column
```

```python
df2['City']
```

```python
df2[['Name', 'City']] # Selecting multiple columns, we need to create list of lists
```

### Creating a new column

```python
df2['Designation'] = ['Doctor', 'Engineer', 'Doctor', 'Engineer']
df2
```

### Removing a column

```python
df2.drop('Designation')
```

We got error because drop checks rows by default, i.e. it didnot find designation in rows

We want python to check columns instead of rows, for that we need to send **'axis=1'** as second parameter in drop. **'axis=0'** is default in drop.

```python
df2.drop('Designation', axis = 1)
```

```python
df2
```

The column don't get deleted. It just gives the view of the dataframe if the columns isn't there.

```python
df2.drop('Designation', axis = 1, inplace = True) # to remove from the table
```

```python
df2
```

```python
df2.drop(['City', 'Salary'], axis = 1) # to remove multiple columns
```

```python
df2.drop(0, axis = 0) # to remove rows
```

## Conditional statement

It means to select based on some conditions 

```python
# people whose age is above 30
df2[df2['Age'] > 30]
```

```python
# people whose age is above 30 and their city must be paris 
df2[(df2['Age'] > 30) & (df2['City'] == 'Paris')]
```
