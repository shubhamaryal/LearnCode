```python
import numpy as np 
import pandas as pd 
```

## Finding missing data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
} # np.nan means null values

df = pd.DataFrame(data)
df
```

```python
df.isna() # to find null values
```

```python
df.isna().sum() # check count of null values, and it gives total null values in each column
```

```python
df.isna().any() # this checks if there is any null value in each column
```

## Removing missing data

Removing missing data works on the basis of rows

```python
df
```

```python
df.dropna()
```

Since all the rows in the DF had NULL values, so everything is removed 

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.dropna()
```

```python
df.dropna(thresh = 3)
# there, the row will only be removed if there are 3 null values in the particular row
```

## Filling missing values

```python
df.fillna(0) # fill null values
```

```python
df
```

We can see that there is no change, and we are just seeing how the DF will look if the method is used, to change it permanently, we have to use **inplace = true**

```python
# to full with different values in each column
values = { 'A': 0 , 'B': 100, 'C': 300, 'D':400 } 
df.fillna(value = values)
```

Selection -> column wise 

Fill -> column wise

drop -> row wise

Delete -> column wise
