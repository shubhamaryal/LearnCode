# Basic Array Creation

```python
arr = np.array([1, 2, 3, 4]) # creates a array
arr
# basically we convert a list into the array using the numpy array method
```

```python
a = [1, 2, 3, 4]
a
```

The ARRAY has "array" written on the output but the list is just covered with "[]". There is no much difference in both.

We can have multiple data type in lists. But we need to have same data type in array.

```python
b = [1, 2, 3.5, 'shubham'] # we have multiple data types here
b
```

```python
np.array(b) # passing the list to convert it into array
```

Here in the above output, we can see dtype='<U32' and it means everything has been converted into unicode or string

```python
c = [1, 2, 4.6] # we have float and interger only this time 
np.array(c)
```

Here, everything is converted into floats

---
There is also vector and matrix which comes from array
- array is 1D
- matrix is 2D (we have multiple rows and columns)

---
Multi-dimensional array is known as matrix. Eg: [[1, 2,3 ],[4, 5,6 ],[7, 8, 9]] 

List doesnot support multi-dimensional array. So we use NumPy to convert into matrix.

```python
l = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
l
```

```python
np.array(l)
```

Everything after 2D i.e. 3D, 4D, etc. are known as Tensors.

# Array generation function

For generating array, firsly we use range function but we write it as "arange" and it is just like for loop, we give start and end point and the end point should be +1 from the end point. We can also provide steps

```python
np.arange(1, 11)
```

```python
np.arange(1, 11, 2)
```

We can also generate "0" using zeros function.

```python
np.zeros(6)
```

Not only array but we can generate matrix too.

```python
np.zeros((4, 6)) # ((rows, columns))
```

Vertical are Columns and Horizontal are Rows
