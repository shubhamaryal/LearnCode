### Broadcasting

```python
l = [10, 20, 30, 40, 50]
arr = np.array(l)
l + 10 
```

Here, when we try to add 10 in all the elements of LIST, we get error, and we need to write a code of loop.

But, we can convert the list to array and perform addition in one line Eg: "arr + 10" ; and it is known as broadcasting.

```python
arr + 10
```

```python
arr2 = np.arange(1, 26).reshape(5 ,5)
arr2
```

### Deep and Shallow copy

```python
a = np.arange(1, 21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a
```

When we do any changed on the array using slice, the main array doesn't change, but in case of list the main array also gets changed, 

And if the default array doesnt change then it is shallow copy and if the main array also gets affected then it is deep copy 

```python
a[4:6] * 10
```

```python
a
```

```python
a 
b = a
b [0] = 99
```

```python
b
```

```python
a
```

This is deep copy.

When we use the slice, it will create a new location in the RAM, but if we copy the array then the pointer gets changed, the location doesn't change, hence the default array also changed, this is called deep copy

If we dont want deep copy to happen while copying the arrarys then we should use copy function.

b = a  [This is deep copy]

b = a.copy()  [This is shallow copy]

```python
a
```

```python
b
```

```python
b = a.copy()
b[0] = 1
```

```python
b
```

```python
a
```

# Advance Array Manipulation

### Stacking array

Stacking means make 2 arrays 1. (Joining 2 arrays)

We can't stack 2 matrix. We can only use stack operations on vectors

There are 3 types of stack:
- horizontal
- vertical
- Column

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
```

```python
np.vstack((a, b)) # Vertical stack
```

```python
np.hstack((a, b)) # Horizontal stack
```

```python
np.column_stack((a, b)) # Column stack
```

### Splitting array

```python
c = np.arange(16).reshape(4, 4)
c
```

There can be 2 splits.
- Horizontal
- Vertical

```python
np.hsplit(c, 2) # Horizontal split
# Syntax: .hsplit(array, number_of_parts)
```

```python
np.hsplit(c, 3) 
```

We can't split in 3 parts, we need to have equal divisions in array

This is for both horizontal and vertical split

```python
np.hsplit(c, 4) 
```

```python
np.vsplit(c, 2) # Horizontal split
# Syntax: .vsplit(array, number_of_parts)
```
