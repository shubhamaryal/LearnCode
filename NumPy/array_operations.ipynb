# Array Operation

### Arithmetic Operation

```python
a1 = np.array([1, 2,3 ,4 ,5])
a2 = np.array([6, 7, 8, 9, 10])
a1 + a2 
```

Here, we have the array with same number of elements, but if we have array with different elements then we will get error

When we are doing any array operation, the size must be same

similarly, we can do *, /, -, //, ** too

### Broadcasting

```python
l = [10, 20, 30, 40, 50]
arr = np.array(l)
l + 10 
```

Here, when we try to add 10 in all the elements of LIST, we get error, and we need to write a code of loop.

But, we can convert the list to array and perform addition in one line Eg: "arr + 10" ; and it is known as broadcasting.

```python
arr + 10
```
