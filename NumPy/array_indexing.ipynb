# Numpy indexing and slicing of vectors

```python
import numpy as np
```

```python
arr = np.arange(11, 21)
arr
```

```python
# The indexing is same as list in array
arr[5] # This will give element in 5th index
```

```python
arr[1:5:2] # [start_index, end_index(it should be +1 of the desired index), step]
```

# Numpy indexing and slicing of Matrix

```python
arr = np.arange(1, 31).reshape(6, 5)
arr
```

```python
arr[0]
```

```python
arr[4] # This gives the row elements, it is how we do row indexing
```

Both row and colum are indexed by integers. 

```python
arr[1, 1] # 1th row, 1th column
```

```python
arr[0, 4] # 0th row, 4th column
```

```python
arr[0:2, 1:3] # Row 0, 1 and Column 1, 2
```

For array slicing, first we need to look that elements to slice

Then, we will look the rows that we want to slice and then go for the columns, the end_index will be +1 here too.

```python
arr[3:6, 3:5]
```

# Boolean Indexing

```python
arr = np.arange(11, 21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
