# Random generation function

We have a RANDOM library which is used to give us random elements in python,

and NUMPY provides us array,

so if both are used together than we can get random arrays

But, when we install NUMPY library, RANDOM library also gets installed automatically 

```python
np.random.rand(5) # here 5 means number of elements we want
```

```python
np.random.rand(10)
```

We get all the values between 0 and 1.

It is because the values are normalized. The normalized values goes from 0 to 1

We can also used standarized values i.e. from -3 to +3 using "**randn**" function

```python
np.random.randn(10)
```

We can also generate integers using **randint**

```python
np.random.randint(6) # Here the 6 denotes "a number between 0 to 6" not "6 numbers"
```

```python
np.random.randint(10, 20, 10) # It means generate 10 number between 10 and 20
# Syntax: (start_point, end_point, numbers_count)
```

# Reshaping and resizing 
It is used when we want to resize the exisiting array

```python
arr = np.arange(1, 31)
arr
```

Here, we have a 1D array or Vector, and we want to resize the array

To resize the array, first we need to call the **reshape()** method.

We need to define the new matrix in the argument. Eg: If we want 5x5 matrix then we will send (5, 5) in the argument.

But, we need to make sure that the new matrix can have all elements of the previous matrix, Eg: if we have 30 elements in the 1D matrix then we can only change it to 6x5 or 5x6. Else, we will get error

```python
arr.reshape(5, 6)
```
