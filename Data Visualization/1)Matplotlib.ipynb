## Object oriented way 

```python
fig = plt.figure()
fig.add_axes([0.1, 0.1, 0.8, 0.8]) 
# 1st and 2nd are the position of the graph in the canvas i.e. x and y
# 3rd and fourth are width and height of the graph in the canvas
# all the values are in %. Eg: 0.8 means 80%
```

```python
fig = plt.figure()
fig.add_axes([0.1, 0.1, 0.3, 0.8])
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.4, 0.3, 0.3])
axis2.plot(x, y)
```

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. We can use the figsize and dpi keyword arguments.

Figsize is a tuple of the width and heigh of the figure in inches dpi is the dots-per-inch (pixel per inch)

```python
fig = plt.figure(figsize = (12, 8), dpi = 100) 
# 12 and 8 are width and height of the canvas
# 12 means (12*100)pixels
# dpi is to make the canvas have more pixels i.e. more dpi means more sharp and clear [100 is default]

axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.4, 0.3, 0.3])
axis2.plot(x, y)
```

## Working with images

```python
import matplotlib.image as mpimg
```

```python
img = mpimg.imread('images.jpeg')
img
```

This array denotes matrix of pixel and each pixel stores color values.

Eg: [R, G, B] -> [255, 0, 0] -> PURE RED

```python
plt.imshow(img)
plt.axis('off')
```

```python
cropped_img = img[50:200, 100:300] # crop: rows 50-200, cols: 100-300
plt.imshow(cropped_img)
plt.axis('off')
```
